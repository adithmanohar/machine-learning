# MACHINE LEARNING

MACHINE LEARNING(ML) is a branch of Artificial Inteligence(AI) that enables  computers to learn from data and make prediction or decisions without beign explicitly programed for every task.

Simple Definition:-

Instead of writing rules manually:
 
     if temparature > 30:
       print('Hot')

In machine learning we provide data and the computer learns the pattern automatically.

Types of ML
 1. # supervised Learning:-
                        The model learns using labeled data.
   
   Two types:- 
              1. Regression  (continues value)
              2. Classification.(group)

Example:
      
      Age       Bought Product 
      25              Yes
      40              No

  Goal: Predict wheather a new customer will buy the product

     -> Algorithms:

    - Linear Regression
        - POLYNOMIAL REGRESSION
             - USED WHEN THE RELATIONSHIP BETWEEN VARIABLES IS NON LINEAR 
                 IT FITS A CURVED LINE INSTEAD OF A STRAIGHT LINE
                     EG- FUEL CONSUPTION
    - Logistic Regression
    - Decision Tree
    - Random Forest 
    - KNN

2. # Un-supervised Learning:-
                              the model learns from unlabeled data

Example:
group customers based on purchasing behaviour

    -> Algorithms

   - K-Means Clustering
   - Hierarchical Clustering
   - DBSCAN

3. # Reinforcement Learning:-
                                the model learns through rewards and penalties

    
    Examples:

           - Self-Driving Cars
           - Game-playing AI

  1. Data:- Information used to train a model. Example- age=[20,25,30,35]
  2. Feature:- Input variable: Example- Age,Salery,Experience (x values- independ)
  3. Target VAriable:- Output Variable to predict.Example- House price

```python
from sklearn import linear_model
from matplotlib import pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
med=data['bedrooms'].median()
data1=data.fillna(med)
data1
```

```python
plt.hist(data1['bedrooms'],edgecolor='black')
plt.xlabel('value')
plt.ylabel('frequency')
plt.title('histogram')
```

```python
plt.scatter(data1['age'],data1['price'])
plt.xlabel('age')
plt.ylabel('price')
plt.title('age vs price')
```

```python
x=data1.drop('price',axis=1)
x
```

```python
y=data1['price']
y
```

```python
model=LinearRegression()
model.fit(x,y) 
```

```python
model.coef_
```

```python
model.intercept_
```

```python
y_pred2=model.predict(x)
y_pred2
```

```python
a=pd.DataFrame({'actualprice':y,
               'predicted price':y_pred2})
a
```

```python
corelation=data1.corr()
corelation
```

```python
sns.heatmap(corelation,annot=True,cmap='coolwarm')
```

```python
model.score(x,y)
```

# evaluation methods

```python
# MSE(mean squred error)
from sklearn.metrics import mean_squared_error
mean_squared_error(y,y_pred2)
```

```python
#MAE(mean absolute error)

from sklearn.metrics import mean_absolute_error
mean_absolute_error(y,y_pred2)
```

```python
#RMSE(ROOT MEAN SQUARE ERROR)
import numpy as np
print(np.sqrt(mean_squared_error(y,y_pred2)))
```
