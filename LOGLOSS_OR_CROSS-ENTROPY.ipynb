## LOG LOSS / CROSS ENTROPY

#### WHAT IS LOG LOSS(CROSS ENTROPY LOSS)

LOG LOSS (CROSS-ENTROPY Loss) is a loss function used to measure how well a classification model predicts the correct class.


- It compares the predicted probability with the actual class label.


#### Why Do We Use Log Loss?

* To evaluate classification models.

* To measure prediction error.

* To improve model performance during training.



#### How Does Log Loss Work?

* If the prediction is correct with high confidence, Log Loss is low.

* If the prediction is wrong or has low confidence, Log Loss is high.


  - GOAL- MINIMIZE LOGLOSS

EXAMPLE

ACTUAL(y)          PREDICTED PROBABILITY(p)         LOG LOSS

    1                       0.99                         VERY LOW(✅)

    1                       0.60                          LOW

    1                       0.20                          HIGH(❌)



HIGHER PROBABILITY FOR THE CORRECT CALSS RESULT IS  LOWER LOG LOSS

```python
from sklearn.metrics import log_loss
```

```python
y_true=[1, 0, 1, 1]
y_pred=[0.9, 0.2, 0.8, 0.7]
```

```python
loss=log_loss(y_true,y_pred)
print('log loss:',loss)
```
