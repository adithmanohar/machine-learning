MULTIPLE PLOTTING USING BAR CHART

```python
x=['shyam','aswin','praveen','anjana','jubil','shibin']
x_range=np.arange(len(x))   #creates array (0,1,2,3,)
wid=0.4  #width of bar in the bar plot

y=[86,75,76,74,72,85]  #height


#xrange- width : shift the bar to the left by the width of bars to avoid overlap
plt.bar(x_range-wid,y,width=wid,label='current weight',color='blue')

y2=[75,75,80,65,70,78]
plt.bar(x_range,y2,width=wid,label='desired weight',color='pink')

plt.xticks(ticks=x_range,labels=x)  #xticks=used to set or customize the label on the x-axis in a plot
#parameter of xticks are: ticks(position on the x-axis where ticks should appear), labels(custom name for each ticks),rotation,fontsize,color

plt.grid()

plt.title('person vs weight diagram')
plt.xlabel('name')
plt.ylabel('weight')
plt.legend(loc='upper left')

plt.show()
```

PIE CHART:-

A PIE CHART IS A CIRCULAR STATISTICAL GRAPH THAT IS DIVIDED INTO SLICES TO ILLUSTRATE NUMERICAL PROPOSTION.

EACH SLICE OF THE PIE REPRESENT A CATEGORY CONTRIBUTION TO WHOLE,MAKING ITS EASY TO COMPARE DIFERENT PARTS OF THE DATA SET.

```python
slices=[30,40,50] #the slices represnt men(30%) women(40%) and kid (50%)
labels=['men','women','kid']  #provide labels for each slices
plt.style.use('dark_background')

plt.pie(slices,labels=labels)   #this function creates the pie chart using the data and labels provided

plt.title('Pie Chart')
```

CUSTOM COLORS AND WEDGE PROPERTIES

 WEDGE PROPERTIES:- In a pie chart, a wedge (or sector) is one of the slices that make up the chart. Each wedge represents a proportionate part of the whole data set. The size of each wedge is determined by the value it represents relative to the total sum of all values. You can customize the appearance of the pie chart wedges using the wedgeprops parameter.

• LINE WIDTH AND EDGE COLOR:- Border line. Example: wedgeprops={'linewidth':2, 'edgecolor':'yellow'} sets the line width to 2 and the edge color to yellow.

• EXPLODS AND SHADOW:- You can use the explode parameter to separate wedges and shadow= True to add a shadow effect

```python
slices=(30,40,20)  # THE SLICE REPRESENT MEN (30%) , WOMEN(40%) AND KID (20%)
labels=['MEN','WOMEN','KIDS']  #PROVIDE LABELS FOR EACH SLICES
col=['yellow','purple','pink']

plt.pie(slices,labels=labels,colors=col,wedgeprops={'linewidth':2,'edgecolor':'red'}) #this function create the pie chart using the data and labels provided

plt.title('pie chart')
plt.style.use('classic')
plt.show()
```

EXPLODES AND SHADOW

```python
slices=(30,40,20)  # THE SLICE REPRESENT MEN (30%) , WOMEN(40%) AND KID (20%)
labels=['MEN','WOMEN','KIDS']  #PROVIDE LABELS FOR EACH SLICES
col=['yellow','purple','pink']

explodes=[0,0,0.2]
plt.pie(slices,labels=labels,colors=col,wedgeprops={'linewidth':2,'edgecolor':'red'},explode=explodes,shadow=True) #this function create the pie chart using the data and labels provided

plt.style.use('classic')
plt.show()
```

STACK PLOT OR AREA CHART 

A stack plot (area chart) is a graph where multiple data series are stacked on top of each other and shown as filled areas

This is useful for visualizing the cumulative value of multiple data series over time or another continous variable

Each dataset is represented by a different color,and the area are stacked on top of each other,showing how each parts contribution


```python
days=[1,2,3,4,5]  #no of days
studying=[6,8,7,6,11] #no of hours spend for studying each day
playing=[1,8,7,6,13]  #no of hours spend playing each day

plt.stackplot(days,studying,playing,colors=['lightblue','pink'])

plt.xlabel=('days')
plt.ylabel=('number of hours')

plt.legend(['studying','playing'],loc="upper left")
```

OUTLIER :-It is a data point that significantly deviates from other observations in the dataset outliers can be unusually high or low compared to the majority of the data.

Any data point that falls outside the whiskers(ie,beyond 1.5*IQR from Q1 or Q3) is considered an outlier and is usually marked with a dot or asteriks.

 Common methods include:
• Z-Score-Identifies outliers by measuring how many standard deviations a data point is from the mean.

• IQR(Inter Quartile Range) - Outliers are detected as values lying below

• Box Plot - Visualizes outliers as points outside the 'whiskers' of the box plot.

• Scatter plot or Histogram - Visual detection of points

BOXPLOT Or BOX AND-WHISKER PLOT

• BOX PLOT or BOX-AND-WHISKER PLOT:- It is a standardized way of displaying the distribution of data based on a five number summary. minimum, first quartile (Q1), median (Q2), third quartile (Q3) and maximum.

It provides a visual represntation of the spread and skeweness of the data, while also highlighting outliers.

• KEY COMPONENTS OF BOX PLOT:

• BOX:- The box itself represents the IQR which is the range between the first quartile (Q1) and third quartile (Q3).
-It contains the middle 50% of the data.
-The line inside the box: This is the median (Q2) or the middle value of the data.


• WHISKERS:- The 'whiskers' extend from the box to the minimum and maximum values that are not considered outliers. 

Whiskers typically cover the range within 1.5 * IQR from Q1 and Q3.

• USES OF A BOX PLOT
• Visualize the spread and skewness of the data.
• Compare distributions between different datasets.
• Identify outliers in the data.
• Summarize large dataset in a compact form.

```python
#Q4: Create a violin plot and box plot side by side comparing 3 datasets


import matplotlib.pyplot as plt
import numpy as np


data1=np.random.normal(100,15,200)
data2=np.random.normal(110,20,200)
data3=np.random.normal(105,10,200)

plt.figure(figsize=(12,5))
# Violin plot
plt.violinplot([data1,data2,data3],positions=[1,2,3])
plt.set_ylabel('Values')
plt.set_title('Violin Plot')
plt.xticks([1,2,3])
plt.xticklabels(['Dataset1','Dataset2','Dataset3'])
# Box
plt.boxplot([data1,data2,data3],labels=['Dataset 1','Dataset 2',
'Dataset 3'])
plt.set_ylabel('Values')
plt.set_title('Box Plot')
```

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import numpy as np
fig, ax = plt.subplots()
x = np.linspace(0, 2*np.pi, 100)
line, = ax.plot(x, np.sin(x))
ax.set_ylim(-2, 2)
def animate(frame):
y = np.sin(y + frame/10)
line.set_ydata(y)
return line,
ani = FuncAnimation(fig, animate, frames=100, interval=50)
plt.show()
```
