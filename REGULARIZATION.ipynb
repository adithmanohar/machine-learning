## Overfitting and Underfitting:-
Overfitting and underfitting are two common problems in machine learning that affect how well a model predicts new data.

### 1. Underfitting -
   A model is underfitting when it is too simple to learn the patterns in the training data.
   (മോഡൽ ഡാറ്റയിലെ യഥാർത്ഥ പാറ്റേൺ ശരിയായി പഠിക്കാത്ത അവസ്ഥയാണ്).

Characteristics:
- High training error ❌
- High testing error ❌
- Poor accuracy on both training and new data

Example:
Suppose you want to predict students' marks based on hours studied.

Actual data:
Hours Marks
  1     20
  2     40
  3     60
  4     80
  5     100

If your model predicts:

Marks = 50
for every student, it ignores the relationship between study hours and marks. This is underfitting.

Causes:
- Model is too simple.
- Too few features.
- Training for too few epochs (in neural networks).
- Too much regularization.
 
How to Fix:
- Use a more complex model.
- Add useful features.
- Train longer.
- Reduce regularization.

ലക്ഷണങ്ങൾ:
- ട്രെയിനിംഗ് ഡാറ്റയിലും മോശം പ്രകടനം.
- പുതിയ ഡാറ്റയിലും മോശം പ്രകടനം.
- മോഡൽ വളരെ ലളിതമാണ് (Too Simple).

ഓർക്കാൻ-
"പഠിക്കേണ്ടത് പോലും ശരിയായി പഠിച്ചില്ല."


### 2. Overfitting -
A model is overfitting when it learns the training data too well, including random noise and outliers.
(മോഡൽ ട്രെയിനിംഗ് ഡാറ്റയിലെ എല്ലാ വിവരങ്ങളും, അതിലെ ചെറിയ പിശകുകളും (Noise) വരെ മനപ്പാഠമാക്കുന്ന അവസ്ഥയാണ് Overfitting.)

ഉദാഹരണം

ഒരു വിദ്യാർത്ഥി കഴിഞ്ഞ വർഷത്തെ ചോദ്യപേപ്പർ മാത്രം മനപ്പാഠമാക്കുന്നു.

Practice Test → 100%
Final Exam → 50%

കാരണം പുതിയ ചോദ്യങ്ങൾക്ക് ഉത്തരം പറയാൻ കഴിയുന്നില്ല.
It memorizes instead of learning the general pattern.

Characteristics:
- Very low training error ✅
- High testing error ❌
- Poor performance on unseen data

Causes:
- Model is too complex.
- Too many features.
- Small training dataset.
- Training for too many epochs.
- No regularization.
 
How to Fix:
- Collect more data.
- Use regularization (L1, L2).
- Use dropout (for neural networks).
- Early stopping.
- Simplify the model.
- Remove irrelevant features.

ലക്ഷണങ്ങൾ:
- Training Accuracy വളരെ കൂടുതലാണ്.
- Testing Accuracy കുറവാണ്.
- പുതിയ ഡാറ്റയിൽ ശരിയായി പ്രവചിക്കാൻ കഴിയില്ല.

ഓർക്കാൻ-
"മനസ്സിലാക്കിയില്ല, മനപ്പാഠമാക്കി."

## Regularization

Regularization is a set of techniques in machine learning that reduces overfitting by discouraging a model from becoming overly complex. The goal is to help the model perform well not only on the training data but also on new, unseen data.

Overfitting ഒഴിവാക്കാൻ ഉപയോഗിക്കുന്ന ഒരു സാങ്കേതിക വിദ്യയാണ് Regularization.

ഇത് മോഡലിനോട് പറയുന്നത്:

"അനാവശ്യമായി എല്ലാം പഠിക്കേണ്ട. പ്രധാനപ്പെട്ട കാര്യങ്ങൾ മാത്രം പഠിക്കൂ."

ഒരു യഥാർത്ഥ ഉദാഹരണം:-

ഒരു അധ്യാപകൻ പരീക്ഷയ്ക്ക് 100 ചോദ്യങ്ങൾ കൊടുക്കുന്നു.

-> Underfitting:- വിദ്യാർത്ഥി 10 ചോദ്യങ്ങൾ മാത്രം പഠിച്ചു. പരീക്ഷയിൽ കുറഞ്ഞ മാർക്ക്.

-> Good Fit:- വിദ്യാർത്ഥി പ്രധാനപ്പെട്ട ആശയങ്ങൾ മനസ്സിലാക്കി പഠിച്ചു. പരീക്ഷയിൽ നല്ല മാർക്ക്.

-> Overfitting:- വിദ്യാർത്ഥി 100 ചോദ്യങ്ങളുടെയും ഉത്തരങ്ങൾ മനപ്പാഠമാക്കി. പരീക്ഷയിൽ പുതിയ ചോദ്യങ്ങൾ വന്നപ്പോൾ ഉത്തരം പറയാൻ   കഴിഞ്ഞില്ല.

- Regularization:-
അധ്യാപകൻ പറയുന്നു:

"എല്ലാം മനപ്പാഠമാക്കണ്ട. ആശയം മനസ്സിലാക്കൂ."

അപ്പോൾ വിദ്യാർത്ഥിക്ക് ഏത് ചോദ്യവും വന്നാലും ഉത്തരം പറയാൻ കഴിയും.

Why is regularization needed?

Imagine you're fitting a curve through data points:

- Underfitting: The model is too simple and misses important patterns.
- Good fit: The model captures the underlying trend.
- Overfitting: The model memorizes the training data, including noise, and performs poorly on new data.

Regularization helps move the model from overfitting toward a better balance.

How does it work?

Most regularization methods add a penalty for large or complex model parameters to the loss function.

-> Common types of regularization:-
### 1. L1 Regularization (Lasso)-
. Encourages some weights to become exactly zero.
. Performs feature selection automatically.
. Produces sparse models.

Example: If you have 100 features but only 10 are important, L1 may set the other 90 weights to zero.


### 2. L2 Regularization (Ridge)
. Shrinks weights toward zero but rarely makes them exactly zero.
. Makes the model more stable.
. Most commonly used regularization technique.

- ഒരു weight-ഉം സാധാരണയായി 0 ആക്കില്ല.

```python
lb=LabelEncoder()
data['town']=lb.fit_transform(data['town'])
data[['town']]
```

```python
x=data.drop('price',axis=1)
x
```

```python
y=data['price']
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3)
```

### LASSO REGRESSION-L1

```python
from sklearn.linear_model import Lasso
```

```python
model=Lasso(alpha=2) #create a lasso regression model.regularization stregthen=2,apply a regularization penalty with strength 2
model.fit(x_train,y_train)


#ALpha parameter controls the strength of L1 regularization penalty 
# #smalL alpha-weak regularization, high risk of overfitting, eg: -0.01,0.1
#medium alpha- eg: 1 or 2
#Large alpha- strong regularization eg: 10-100, may under fit the data.
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
model.score(x_test,y_test)
```

```python
r2=r2_score(y_test,y_pred)
r2
```
