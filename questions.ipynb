```python
#Create a violin plot using seaborn with three different categories showing the distribution of values. Use hue parameter to differentiate categories by color.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Create sample data
np.random.seed(42)

data = pd.DataFrame({
    "Category": np.repeat(["A", "B", "C"], 100),
    "Value": np.concatenate([
        np.random.normal(50, 10, 100),
        np.random.normal(65, 12, 100),
        np.random.normal(80, 8, 100)
    ]),
    "Group": np.tile(np.repeat(["X", "Y"], 50), 3)
})

# Create violin plot
plt.figure(figsize=(8, 6))
sns.violinplot(
    data=data,
    x="Category",
    y="Value",
    hue="Group",
    palette="Set2",
    split=True
)

plt.title("Violin Plot of Value Distribution by Category and Group")
plt.xlabel("Category")
plt.ylabel("Value")
plt.legend(title="Group")
plt.show()
```

```python
#Create a histogram of normally distributed data (1000 samples, mean=100, std=15) with a kernel density estimate (KDE) overlay. Customize colors and labels.

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Generate normally distributed data
np.random.seed(42)
data = np.random.normal(loc=100, scale=15, size=1000)

# Create histogram with KDE
plt.figure(figsize=(10, 6))

sns.histplot(
    data,
    bins=30,
    kde=True,
    color='skyblue',
    edgecolor='black',
    alpha=0.7,
    line_kws={'color': 'darkred', 'linewidth': 2}
)

# Customize plot
plt.title('Histogram of Normally Distributed Data with KDE', fontsize=14)
plt.xlabel('Value', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```

```python
#Create a categorical plot (bar plot or count plot) using seaborn with data containing multiple categories. Add error bars using the ci parameter.



import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Create sample data
np.random.seed(42)

data = pd.DataFrame({
    "Category": np.repeat(["A", "B", "C", "D"], 30),
    "Score": np.concatenate([
        np.random.normal(70, 8, 30),
        np.random.normal(80, 10, 30),
        np.random.normal(75, 7, 30),
        np.random.normal(85, 9, 30)
    ])
})

# Create bar plot with confidence interval error bars
plt.figure(figsize=(8, 6))

sns.barplot(
    data=data,
    x="Category",
    y="Score",
    ci=95,          # 95% confidence interval
    palette="Set2"
)

plt.title("Average Score by Category with 95% Confidence Intervals")
plt.xlabel("Category")
plt.ylabel("Average Score")

plt.show()
```

```python
#Create a scatter plot where point colors represent the magnitude of a third variable.Include a colorbar to show the scale. Use a continuous colormap.


import numpy as np
import matplotlib.pyplot as plt

# Generate sample data
np.random.seed(42)

x = np.random.rand(100) * 100
y = np.random.rand(100) * 100

# Third variable that controls point color
z = np.sqrt(x**2 + y**2)

# Create scatter plot
plt.figure(figsize=(8, 6))

scatter = plt.scatter(
    x,
    y,  
    c=z,                # Color based on z values
    cmap='viridis',     # Continuous colormap
    s=80,               # Marker size
    alpha=0.8
)

# Add colorbar
cbar = plt.colorbar(scatter)
cbar.set_label('Magnitude (z)', fontsize=12)

# Labels and title
plt.title('Scatter Plot with Color-Mapped Third Variable')
plt.xlabel('X Values')
plt.ylabel('Y Values')

plt.grid(True, alpha=0.3)
plt.show()
```

```python
#Create a plot with 3 different line plots and customize the legend with custom location, frame, shadow, and fontsize parameters.

import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(0,10,200)

y1=x
y2=x**2
y3=np.sqrt(x)


plt.figure(figsize=(10,6))

plt.plot(x,y1,color='blue',linestyle='-',linewidth=2,label='y=x')
plt.plot(x,y2,color='red',linestyle='--',linewidth=2,label='y=x²')
plt.plot(x,y3,color='green',linestyle=':',linewidth=2,label='y=√x')


plt.legend(
    loc='upper left',  
    fontsize=12,        
    frameon=True,       
    shadow=True,        
    title='Functions'
)

plt.title('Multiple Line Plots with Customized Legend')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True,alpha=0.3)

plt.show()
```

```python
#Create a pair plot using seaborn from a dataset with at least 4 numerical columns. Use hue parameter to color by a categorical variable.
iris = sns.load_dataset("iris")


sns.pairplot(
    iris,
    vars=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ],
    hue="species",
    palette="Set2",
    diag_kind="hist"
)

plt.suptitle(
    "Pair Plot of Iris Dataset",
    y=1.02,
    fontsize=16
)

plt.show()
```

```python
#Create multiple plots (at least 3) using different seaborn color palettes. Compare 'husl', 'Set2', and 'pastel' palettes.

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample data
data = pd.DataFrame({
    "Category": ["A", "B", "C", "D", "E"],
    "Value": [25, 40, 30, 50, 35]
})

# Create subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Palette 1: husl
sns.barplot(
    data=data,
    x="Category",
    y="Value",
    palette="husl",
    ax=axes[0]
)
axes[0].set_title("Palette: husl")

# Palette 2: Set2
sns.barplot(
    data=data,
    x="Category",
    y="Value",
    palette="Set2",
    ax=axes[1]
)
axes[1].set_title("Palette: Set2")

# Palette 3: pastel
sns.barplot(
    data=data,
    x="Category",
    y="Value",
    palette="pastel",
    ax=axes[2]
)
axes[2].set_title("Palette: pastel")

plt.tight_layout()
plt.show()
```
