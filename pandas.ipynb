Here, we replace occurrence of alphabets (both upper and lower) in the 'temperature' column with an empty string.

Replace any occurrence of alphabets in the 'event' column with an empty string.

The replace() method will remove all alphabetic character from the specified columns.

regex=True This parameter indicates that the replacement values are specified using regular expression.

This means that instead of exact string matches, pandas will look for patterns that match the specified values.


```python


data.replace('-99999',np.nan)


data.replace({'temperature':['-99999','40'],'windspeed':'-88888','event':'0'},np.nan)


data.replace({'-99999':np.nan,'-88888':np.nan,'0':'rain'})


data.replace({'temperature':'[A-Za-z]','windpeed':'[A-Za-z]','event':'[A-Za-z]'},'',regex=True)


data.replace(['Rain','Sunny','Cloudy','Snow'],[100,101,102,103])
```

when you apply groupby to a categorical column in a dataframe, it returns a Groupßy object.
This object allows you to perform various operations on the grouped data.

CONCATINATION

```python
tcs={'id':[101,102,103,104],
'name':['max','Dipin', 'Jose','rahul'],
}
wipro={'id': [101,102, 103,104],
'name': ['Dev', 'Sreenath', 'Sreerag', 'Shafeel'],
'age': [24, 24, 27, 26]
}
```

```python
a=pd.DataFrame(tcs)
a

b=pd.DataFrame(wipro)
b

pd.concat([a,b])  #JOIN TWO DATAFRAMES
```

```python
pd.concat([a,b],ignore_index=True)

#get common index after concatination
```

```python
pd.concat([a,b],keys=['tcs','wipro'])
#vertically arraging
```

```python
pd.concat([a,b],axis=1)

#horizontly arranging
```

CROSS TABLE OR CONTIGENCY TABLE

DISPLAY TEH FREQUENCY DISTRIBUTION OF VARIABLE

```python
data={'gender':['male','female','female','male','male'],
      'handedness':['right-handed','left-handed','right-handed','right-handed','left-handed']
      
    }
#creating cross tabulation
df=pd.DataFrame(data)
crosstab=pd.crosstab(df['gender'],df['handedness'])
crosstab
#it takes a two columns on the data frame df- gender and handedness.the function count the occurance
#the table shows the count of each combination of gender and handedness.
```
