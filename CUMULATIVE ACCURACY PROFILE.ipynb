# CAP(CUMULATIVE ACCURACY PROFILE)

## WHAT IS CAP?

CAP(CUMULATIVE ACCURACY PROFILE) IS A graphical method used to evaluate the perfomance of a binary classification model

it shows how well the model identifies positive cases compared to a random model

## why do we use CAP

- to evaluate classification models
- to compare a model with a random model
- to mesure the model prediction accuraacy visually

## How does CAP Work?

- the X-axis represent the cumulative percentage of data
- the Y-axis represent the cumulative percentage of data of positive cases captured
- A better model produces a curve closer to the ideal model.

## CAP CURVES

- RANDOM MODEL:- Straight diagonal line.
- IDEAL MODEL:- Reaches 100% positive very quickly
- TRAINED MODEL:- Lies between the Random and Ideal Models

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# X-axis values ( CUMULATIVE % OF DATA)

x=np.array([0,20,40,60,80,100])
```

```python
# IDEAL MODEL VALUES
 
ideal=np.array([0,80,100,100,100,100])
```

```python
# TRAINED MODEL VALUES 

model=np.array([0,55,75,90,97,100])
```

```python
# RANDOM MODEL VALUES(DIAGONAL LINE)
random=x
```

```python
# plot ideal model

plt.plot(x,ideal,label='Ideal Model')

#plot Trained model

plt.plot(x,model,label='Trained model')

# plot the random model

plt.plot(x,random,'--',label='Random model')


plt.xlabel('cumulative % of data')
plt.ylabel('cumlative % of positive cases')

plt.title('CAP CURVE')
```

### Output Explanation
* Ideal Model captures almost all positive cases with fewer samples.
* Random Model follows a straight diagonal line.
* A good model lies closer to the Ideal Model than the Random Model.

## Real-Life Example
A bank uses a CAP curve to evaluate a loan approval model. If the model identifies most good customers early, it is considered a good model.


# Summary
* CAP stands for Cumulative Accuracy Profile.
* It is used to evaluate binary classification models.
* It compares the Model, Random Model, and Ideal Model.
* A curve closer to the Ideal Model indicates better performance.
* CAP helps visualize the effectiveness of a classification model.
