```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np
```

```python
a=np.array([[1],[2],[3],[4]])
b=np.array([2,4,6,8])
model=LinearRegression()
model.fit(a,b)
c=model.predict(a)
print(model.predict([[5]]))
```

```python
plt.scatter(a,b)
plt.plot(a,c,color='red')
plt.xlabel('a')
plt.ylabel('b')
```
