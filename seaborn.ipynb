SEABORN

Seaborn is a python data visualization library built on top of matplotlib that provides a high-level interface for creating informative and attractive statistical graphics.

It simplifies the creation of complex plots with minimal code, offering default aesthetics and more advanced options for styling and handling complex datasets.

Seaborn is specifically designed to work with pandas data structures (eg: dataframes) and includes tools for easily visualizing relationship between multiple variable and performing statistical analysis

Seaborn Special:-

Violin plot
Heatmap (better version)
Pairplot
Countplot
KDE plot

2.BAR PLOT

BAR PLOT ALWAYS AGGREGATE DATA 
DEFAULT=MEAN THATS WHY SEABORN DIFFERENT FROM MATPLOTLIB

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips')
df
```

```python

sns.barplot(data=df,x='day',y='total_bill')

#shows average  total_bill for each day
```

```python
sns.barplot(data=df,x='day',y='total_bill',hue='sex')
```

```python
import numpy as np
```

```python
sns.barplot(data=df,x='day',y='total_bill',estimator=np.sum)
#function used to calculate value(default=mean)
#now shows the sum instead of average
```

```python
sns.barplot(data=df,x='day',y='total_bill',errorbar=('ci',95))

#error bar (deafult 95%) 95%ci(confidence intervel):- difference between calculated average and orginal average
```

```python
#use ci=none to remove

sns.barplot(data=df,x='day',y='total_bill',ci=None)
```

```python
sns.barplot(data=df,x='day',y='total_bill',n_boot=500)
#number of bootstrap samples=taking 500 samples and calculate mean samples (500 mean will get ) using these 500 mean,error bar will get
```

```python
sns.barplot(data=df,x='day',y='total_bill',palette='viridis',saturation=0.5)
```

```python
sns.barplot(data=df,x='day',y='total_bill',color='red')
```

```python
#custom order of categories

sns.barplot(data=df,x='day',y='total_bill',
            order=['Sun','Sat','Thur','Fri'])

```

```python
sns.barplot(data=df,x='day',y='total_bill',
            hue='sex',hue_order=['Female','Male'])
```

```python
sns.barplot(data=df,x='total_bill',y='day',orient='h')
```

```python
sns.barplot(data=df,x='day',y='total_bill',
            hue='sex',dodge=False)
```

```python
sns.barplot(data=df,x='day',y='total_bill',linewidth=2,edgecolor='black')
```

```python
sns.barplot(data=df,x='day',y='total_bill',errcolor='red',errwidth=2)
```

```python
sns.barplot(data=df,x='day',y='total_bill',capsize=0.2)
```

```python
sns.barplot(data=df,
                        x='day',
                        y='total_bill',
                        hue='sex',
                        estimator=np.mean,
                        ci=95,
                        palette='coolwarm',
                        capsize=0.2,
                        errwidth=2,
                        edgecolor='black',
                        linewidth=1,
                        saturation=0.8)
                        
plt.title('bar plot example- Tips Dataset')
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data1=np.random.normal(loc=0,scale=1,size=1000)
data2=np.random.normal(loc=1,scale=1,size=1000)
sns.kdeplot(data1,fill=True,label='data1')
sns.kdeplot(data2,fill=True,label='data2')



```
