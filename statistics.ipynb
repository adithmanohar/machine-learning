# SKEWNESS

 - IT IS A MEASURE OF THE ASYMMETRY OF THE PROBABILITY DISTRIBUTION OF A REAL-VALUES RANDOM VARIABLE ABOUT ITS MEAN
 - IT INDICATE WHEATHER THE DATA POINTS ARE SKEWES TO THE LEFT (NEGATIVE SKEW )OR TO THE RIGHT(POSITIVE SKEW)

Positive skew (right skew): The right tail (lower values) is longer or fatter than the left tail.
This means that the bulk of the values (including the median) lie to the left of the mean.

*Negative skew (Left skew): The left tail (lower) is longer or fatter than the right tail.
This means that the bulk of the values (including the median) lie to the right of the mean.

* Zero skew: The data is symmertrically distributed around the mean, meaning the left and right tails are mirror.

# KURTOSIS

kurtosis measures the taildness of a distribution or how much data lies in the tails compared to the center.

* High kurtosis (Leptokurtic):-
Heavy tails and a sharp peak. More outliers.

* Low kurtosis (Playkurtic):-
Light tails and a flatter peak. Fewer outliers.

* Normal kurtosis (Mesokurtic):-
The distribution resemble a normal distribution.


Skewness and kurtosis are used to check assumption of normality in statisticsl models.
Both measures help in understanding the distribution shape which is essential for data analysis and hypothesis testing.
