IN UNSUPERVISED LEARNING ,THE TRAINING DATA DOES NOT CONTAIN OUTPUT LABELS

The model groups similar data points and discovers the hidden pattern without human supervision

# Topic 4: K-Means Clustering



## What is K-Means Clustering?

K-Means is a popular clustering algorithm that groups similar data points into a predefined number of clusters (K).

The algorithm groups data based on the distance between data points and the cluster center (centroid).

## What is K?

K- represents the number of clusters we want to create.


## Example:

K = 2 → 2 Clusters

K = 3 → 3 Clusters

K = 4 → 4 Clusters

## How Does K-Means Work?

Step 1: Choose the value of K

Step 2: Select Initial Centroids

Step 3: Assign Data Points

Step 4: Update Centroids

Step 5: Repeat Until No Changes




## Simple Diagram
Step 1

●   ●      ▲   ▲

      K = 2

Step 2

Cluster 1      Cluster 2

● ● ●          ▲ ▲ ▲
     ●              ▲



## Advantages
Simple and easy to understand.
Works well for large datasets.
Fast clustering algorithm.

## Limitations
The value of K must be chosen before training.
Sensitive to outliers.
Works best with spherical clusters.

## Real-Life Example
An online shopping company groups customers into different categories based on their purchasing behavior using K-Means Clustering.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

```

```python
x=[
    [1,2],
    [2,3],
    [3,3],
    [8,8],
    [9,8],
    [8,9]
]


model=KMeans(n_clusters=2,random_state=42)

#train model
model.fit(x)

#predict clusters
labels=model.labels_
print(labels)
```

```python
plt.scatter(
    [i[0] for i in x],
        [i[1] for i in x],
        c=labels,
        s=100

)

plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    marker='x',
    s=200


)

plt.title('k_means clustering')
plt.show()
```

## Summary

- K-Means is a popular clustering algorithm.
- It groups similar data into K clusters.
- Each cluster has a centroid.
- Data points are assigned to the nearest centroid.
- The process repeats until the clusters become stable.

# TOPIC 5 DBSCAN CLUSTERING 


## WHAT IS DBSCAN?

DBSCAN stands for DENSITY-BASED SPATIAL CLUSTERING OF APPLICATION WITH NOISE

IT groups closely packed data points and identifies isolated point as noise

## important parameter

eps

eps defines the maximum distance beteween two points to be considered negibours

min_samples

min_samples  specifies the minimum number of points required to form a dense region

### How dbscan works

1. choose eps and min_samples 
2. find neighbouring point
3. create clusters from dense region
4. identify isolated point as noise

## simple daiagram

Cluster 1              Cluster 2

 ● ● ●                  ● ●
 ● ● ●                ● ● ●
   ●                      ●

                    ×
                  Noise  

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
```

```python
x=[
    [1,2],
    [2,2],
    [2,3],
    [8,8],
    [8,9],
    [9,8],
    [20,20]

]
```

```python
model=DBSCAN(eps=2,min_samples=2)

labels=model.fit_predict(x)
print(labels)
```

```python
plt.scatter(
    [i[0] for i in x],
    [i[1] for i in x],

    c=labels
)
plt.title('db scan clustering')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

Combine frequent items to create larger itemsets.

#### Step 3: Remove Infrequent Itemsets
Itemsets below the minimum support are removed.

#### Step 4: Generate Association Rules
Association rules are created from the frequent itemsets.

# Topic 9: Principal Component Analysis (PCA)
### What is PCA?
PCA stands for Principal Component Analysis.
PCA is a dimensionality reduction technique used to reduce the number of features while preserving important information.


### Why Do We Use PCA?
* To reduce the number of features.
* To simplify complex datasets.
* To visualize high-dimensional data.
* To reduce computation time.


### How PCA Works
Step 1: Standardize the Data

Features are brought to a similar scale.

Step 2: Find Principal Components

PCA finds new directions that contain the most important variation in the data.

Step 3: Select Components

The most important components are selected.

Step 4: Transform the Data

The original features are transformed into the selected principal components.



### Simple Diagram
Original Features
F1   F2  F3  F4
 |_____|___|____|
       |
      PCA
       |
Principal Components

    PC1    PCZ

### Example:

4 Features
    |
   PCA
    |
2 Components



```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
# Load dataset
iris = load_iris()
x=iris.data

# standardize data
scaler=StandardScaler()
x_scaled = scaler.fit_transform(x)


```

```python
# Apply PCA
pca = PCA(n_components=2)
x_pca=pca.fit_transform(x_scaled)
print("original Shape:",x.shape)
print("PCA Shape:", x_pca.shape)
```

### EXPLANATION 

THE orginal iris dataset has 150 samples and 4 feature .PCA convert the 4 feature into 2 principle component

### VISUALIZING PCA

```python
import matplotlib.pyplot as plt
plt.scatter(x_pca[:,0],x_pca[:,1])
plt.xlabel("principle component 1")
plt.ylabel("principle component 2")
plt.title("PCA Visualization")
```

## Important Point
Principal components are new features created from the original features.
PC1 usually contains the highest amount of variation, followed by PC2, PC3, and so on.


## Real-Life Example
Suppose a dataset has 50 features. PCA can reduce them to 5 important components. making the dataset easier to analyze.
## Summary
* PCA stands for Principal Component Analysis.
* It is a dimensionality reduction technique.
* It creates new features called principal components.
* It reduces the number of features while retaining important information.
* PC1 generally captures the highest variation
