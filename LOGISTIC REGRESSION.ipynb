# LOGISTIC REGRESSION   (SUPERVISED - CLASSSIFICATION)


Unlike linear regression, which predict contineous value ,Logistics regression predict probability that an instance  belongs to a particular class

-it uses the sigmoid(logistics)function to convert the linear output to probability  beteween 0 AND1.

##### sigmoid function(logistic function)

- The logistic function(or sigmoid function) convert the log odds back to a probability value between 0 and 1

σ(z)=1/1+e - z1

decision Rule :-
if probability > 0.5 -> calss=1
if probability <= 0.5 -> calss=0

MAXIMUM LIKELIHOOD ESTIMATION(MLE):

The parameter (coifficients) of the logistics regression model are estimated using MLE,which finds the parameter value that maximise the likelihood of observing the given data

അതുകൊണ്ടാണ് Logistic Regression ഉപയോഗിക്കുന്നത്.



Logistic Regression എങ്ങനെ Probability ഉണ്ടാക്കുന്നു?

ഇത് Sigmoid Function എന്ന ഗണിത ഫംഗ്ഷൻ ഉപയോഗിക്കുന്നു.

Sigmoid Function എന്ത് ചെയ്യുന്നു?
σ(z) = 1/1+e - z1

ഏത് വലിയ സംഖ്യയും എടുത്താലും Output 0 മുതൽ 1 വരെ മാത്രമാക്കും.

അതുകൊണ്ട് Probability ലഭിക്കുന്നു.


Decision Rule

Probability ≥ 0.5

➡️ Class = 1 (Yes / Pass / Spam)

Probability < 0.5

➡️ Class = 0 (No / Fail / Not Spam)

# CONFUSION MATRIX
It is a useful tool for evaluating the perfomance of a classification model.
It provides a summary of the prediction results on a classification problem.
The matrix itself is a table with 4 different combinations of predicted and actual values. Here is a quick breakdown:
The confusion matrix tells us how well the model works and gives us a matrix or table as output.
- sometimes this type of structure is called the error matrix.
- The matrix is a summary of the results of the predictions.
It shows how many predictions were right and how many were wrong.




- True Positive  (TP): The model correctly predicted the positive class
- True Negative  (TN): The model correctly predicted the negative class
- False Positive (FP): The model incorrectly predicted the positive class (Type1 error)
- False Negative (FN): The model incorrectly predicted the negative class (Type2 error)

```python
#Confusion matrix
cm=confusion_matrix(y_test, y_pred)
print(f"Confusion matrix: {cm}")
```

```python
sns.heatmap(cm,annot=True,cmap='Blues',xticklabels=['predicted 0','predicted 1'],yticklabels=['actual 0','actual 1'])

plt.xlabel('predicted')
plt.ylabel('actual value')
plt.title('confusion matrix heat map|')
```

# CLASSIFICATION REPORT
The classification_report function in scikit-learn is a handy tool for evaluating the performance of a classification model.
It provides a detailed summary of key metrics for each class, including:\

#### Precision:
The ratio of correctly predicted positive observations to the total predicted positives.
It indicates the accuracy of positive predictions.

#### Recall:
The ratio of correctly predicted positive observations to all observations in the actual class.
It measures the model's ability to find all positive samples.

#### F1-Score:
The weighted harmonic mean of precision and recall.
It provides a balance between precision and recall, especially useful when you need to consider both false positives and false in

The number of actual occurrences of each class in the dataset.
#### Macro Average:
It is the normal average of these metrics.

#### Weighted Average:
Weighted average of the metrics which differs from the macro average for unbalanced datas


```python
print(classification_report(y_test, y_pred))
```
