#  ROC AND AUC 

### WHAT IS ROC

ROC(RECIEVER OPERATING CHARACTERISTIC) is a graph used to evaluate the perfomance of a binary classification model.

it shows how well the model distinguishes between positive and negative classes

## WHAT IS AUC?

AUC(AREA UNDER ROC CURVE) is the area under the ROC curve

it measure the overall perfomance of the classification model

### WHY DO WE USE ROC AND AUC 

- To evaluate binary classification model 
- To compare different classification model
- To measure the model ability to separate classes

## UNDERSTANDING ROC

- X-axis --> False Positive Rate(FPR)
- Y-axis --> True Positive Rate(TPR)

## UNDERSTANDING AUC SCORE

AUC SCORE                    PERFOMANCE

    1.0                     PERFECT MODEL

    0.9-1.0                  EXCELLENT

    0.8-0.9                     GOOD

    0.7-0.8                     FAIR

    0.5                      RANDOM GUESS

    <0.5                      POOR MODEL

```python
from sklearn.metrics import roc_curve,auc
import matplotlib.pyplot as plt
```

```python
y_true=[0, 0, 1, 1]
y_score=[0.1, 0.4, 0.35, 0.8]
```

```python
fpr,tpr,threshold=roc_curve(y_true,y_score)

roc_auc=auc(fpr,tpr)
```

```python
print('Auc Score =',roc_auc)
```

```python
plt.plot(fpr,tpr,label='ROC CURVE')
plt.plot([0,1],[0,1],'--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC CURVE')
plt.legend()
```
