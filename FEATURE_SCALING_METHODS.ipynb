## FEATURE SCALING METHODS

- FEATURE SCALING is the process of bringing all numerical feature to a similar range
- it helps machine learning algorithms perfoem better and converg faster

## NORMALIZATION(Min-Max Scaling)

- Normalization scales the value beteween 0 and 1

  - EXAMPLE:-
  
              - 20 --> 0.2

              - 50 --> 0.5

              - 100 --> 1.0


Smallest value becomes 0 and largest become 1

## STANDARDIZATION

STANDARDIZATION transfrom the data to have a mean of 0 and STANDARD DEVIATION OF 1

- after STANDARDIZATION the data is centered around zero

### when to use

- use Normalization when the data has a fixed range

- use STANDARDIZATION when the data contain outliers or follows a Normal Distribution

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
import numpy as np
```

```python
data= np.array([[10],[20],[30],[40],[50]])
```

```python
#Normalization

minmax=MinMaxScaler()
print(minmax.fit_transform(data))
```

```python
#STANDARDIZATION

standard=StandardScaler()
print(standard.fit_transform(data))
```
