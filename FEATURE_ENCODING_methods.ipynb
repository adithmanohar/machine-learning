## FEATURE ENCODING METHODS

Feature Encoding converts categorical data into numerical form

## LABEL ENCODING

LABEL ENCODING assign a unique number to each category 

# EXAMPLE

 - RED--> 0
 - BLUE--> 1
 - GREEN--> 2

IT IS SUITABLE FOR ORDINAL CATEGORIES

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python

colors=['red','blue','green','red']

le=LabelEncoder()
print(le.fit_transform(colors))


```

## ONE HOT ENCODING

ONE-HOT ENCODING CREATES A SEPARATE BINARY COLUMN FOR EACH CATGORY



EXAMPLE



COLOUR          RED     BLUE     GREEN


RED             1       0         0


BLUE            0       1         0

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python
colors=['red','blue','green','red']


df=pd.DataFrame({'colors':colors})
```

```python
print(pd.get_dummies(df))
```

### method 2

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
import pandas as pd
```

```python
colors=['red','blue','green','red']
```

```python
#label encoding

le=LabelEncoder()
print(le.fit_transform(colors))
```

```python
#one-hot encoding

df=pd.DataFrame({'color':colors})
df
```

```python
encoder=OneHotEncoder(sparse_output=False)
encoded=encoder.fit_transform(df)
print(pd.DataFrame(encoded))
```
