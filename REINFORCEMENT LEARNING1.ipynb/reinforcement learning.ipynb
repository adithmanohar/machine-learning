## Part 1: Introduction to Reinforcement Learning

### What is Reinforcement Learning?
Reinforcement Learning (RL) is a type of Machine Learning where an agent learns by interacting with an environment and receiving feedback.

The agent learns which actions are good and which actions are not good based on rewards and penalties.

## Main Components of Reinforcement Learning

There are primarily five components in Reinforcement Learning.


1. Agent
The agent is the leamer or decision-maker.

Example: Robot, Game-playing Al

2. Environment
The environment is the world in which the agent operates.

Example: Game world, Room, Road

3. State
A state represents the current situation of the agent or environment.

Example:

4. Action
An action is a decision taken by the agent.

Example:

5. Reward
A reward is feedback received by the agent after taking an action.

Good Action --> +10 Reward

Bad Action  → -5 Penalty

 # Topic 2: Important RL Terminology

## 1. Episode

An episode is one complete sequence of interaction between the agent and the environment, from the beginning to the end.

### Example:
From the start of a game until a win or loss → one episode

start-->Actions-->Rewards-->Game Ends-->Episode Complete

## 2. Policy

A policy defines how an agent chooses an action for a given state.

### Example:
State: Robot is near a wall --->Policy--->Action: Turn Right

## 3. Reward

A reward is feedback received after an action.

Good Action → +10
Bad Action → -5

The agent is trying to maximize the reward.

## 4. Value

Value represents how good a state is based on the expected future rewards.

State A + High Future Rewards + High Value

state B → Low Future Rewards + Low Value


## 5. Q-Value
Q-Value represents how good an action is when taken in a particular state.

### Example:

    State             Action       Q-Value              
    
       A                   Right      8
       
       A                     Left       2
    
                       
In State A:

Right - Q-Value - 8
Therefore, there is a higher probability of choosing Agent Right.


# VALUE VS Q VALUE

VALUE                              Q-VALUE

Evaluate a state                   Evaluate state actio pair

How good is this state?            How good is this action in this state?

## 6. Exploration

Exploration means trying new actions to discover whether they give better rewards.

### Example:

Robot normally + Right

New attempt     → Left

## 7. Exploitation

Exploitation means choosing an action that is already known to give a good reward

Previous experience:
Right → High Reward

Current decision:
Right → Choose again

# REINFORCEMENT LEARNING ALGORITHMS

### WHAT ARE  REINFORCEMENT LEARNING ALGORITHMS?

REINFORCEMENT LEARNING ALGORITHMS are methods that help an agent learn the best action through interaction,reward and experence

# Common Reinforcement Learning Algorithms

## 1. Q-Learning
Q-Learning is a value-based Reinforcement Learning algorithm that learns the best action to take in each state.
It stores the learned action values in a Q-table.

### Q-Table           


                       Actions
                     Left  Right
state   0              5     10                 
state   1              8      4
state   2              3      12

The agent generally chooses the action with the higher Q-value.

## 2. SARSA

SARSA is a value-based algorithm that learns from the action actually taken by the agent.
SARSA stands for State, Action. Reward, State, Action.
state-->Action-->Reward-->next State-->Next Action

## 3. Deep Q-Network (DQN)

DQN combines Q-Learning with a Neural Network.

Instead of using a simple Q-table, DQN uses a neural network to estimate action values.

state-->Neural Network-->Q-values-->Best Action

Q-LEARNING VS SARSA VS DQN

Algonthim              Main idea

Q-Learning             Leams the best action using Q-values

SARSA                  Leams from the action actually taken

DQN                    Uses a Neural Network to estimate


## BASIC Q LEARNING EXAMPLE

```python
import numpy as np

# Q-table: 3 states, 2 actions
q_table = np.zeros((3, 2))

learning_rate = 0.2
discount = 0.9

state = 0
action = 1
reward = 10
next_state = 1

# Q-learning update formula
q_table[state, action] = q_table[state, action] + learning_rate * (
    reward
    + discount * np.max(q_table[next_state])
    - q_table[state, action]
)

print(q_table)
```

### Explanation
Initially, all Q-values are 0

The agent takes an action and receives a reward of 10.

The Q-value of that action is updated based on the reward and learning parameters.

### Important Terms

Learning Rate (a)

The ratio of how much old knowledge to discard, and how much new knowledge to trust

Discount Factor (y)

Discount factor determines how much importance is given to future rewards.
