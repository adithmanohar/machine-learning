## POLYNOMIAL REGRESSION
USED WHEN THE RELATIONSHIP BETWEEN VARIABLES IS NON LINEAR 

IT FITS A CURVED LINE INSTEAD OF A STRAIGHT LINE
                
                 EG- FUEL CONSUPTION

```python
import pandas as pd
import numpy as np
from sklearn import linear_model
from matplotlib import pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
x=data[['X']]
y=data[['y']]
```

```python
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x)
```

```python
model=LinearRegression()
```

```python
model.fit(x_poly,y)
```

```python
y_pred=model.predict(x_poly)
```

```python
new_x=[[11]]
new_x_poly=poly.transform(new_x)
prediction=model.predict(new_x_poly)
print('prediction for x=11:', prediction[0])
```

```python
plt.scatter(x,y,color='blue',label='actual data')
plt.plot(x,y_pred,color='red', label='polynomial regression')

plt.xlabel('X')
plt.ylabel('y')
plt.title('POLYNOMIAL REGRESSION')
plt.legend()
```
